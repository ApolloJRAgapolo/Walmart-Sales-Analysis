```python
# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper boundaries
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

# Identify potential outliers
outlier_mask = (
    (df["Weekly_Sales"] < lower_bound) |
    (df["Weekly_Sales"] > upper_bound)
)

# Get potential outlier records
potential_outliers = df[outlier_mask]

# Count potential outliers
print("Number of potential outliers:")
print(potential_outliers.shape[0])
```

```python
# Identify unusually high sales
high_sales_outliers = df[df["Weekly_Sales"] > upper_bound]

# Identify unusually low sales
low_sales_outliers = df[df["Weekly_Sales"] < lower_bound]

print("High sales outliers:", high_sales_outliers.shape[0])
print("Low sales outliers:", low_sales_outliers.shape[0])
```

```python

# Count high and low outliers per store
high_counts = high_sales_outliers.groupby("Store").size()
low_counts = low_sales_outliers.groupby("Store").size()

# Get total records per store
store_records = df.groupby("Store").size()

# Combine the counts into one table
outlier_summary = pd.DataFrame({
    "High_Outliers": high_counts,
    "Low_Outliers": low_counts
}).reindex(store_records.index).fillna(0).astype(int)

# Calculate total outliers
outlier_summary["Total_Outliers"] = (
    outlier_summary["High_Outliers"] +
    outlier_summary["Low_Outliers"]
)

# Add total records
outlier_summary["Total_Records"] = store_records

# Calculate percentage of flagged records
outlier_summary["Outlier_Rate_%"] = (
    outlier_summary["Total_Outliers"] /
    outlier_summary["Total_Records"] * 100
).round(2)

# Display the results
outlier_summary.sort_values(
    "Total_Outliers",
    ascending=False
)
```

```python
# import the required libraries
from pathlib import Path

# create a folder for the exported charts
chart_folder = Path("/content/walmart_python_charts")
chart_folder.mkdir(parents=True, exist_ok=True)

# remove old chart images to avoid duplicate exports
for old_chart in chart_folder.glob("*.png"):
    old_chart.unlink()

# prepare a counter to name each chart
chart_count = 0
saved_figures = set()
saved_figure_objects = []

# keep the original show function
original_show = plt.show

# save each chart before displaying it
def save_chart_before_display(*args, **kwargs):
    global chart_count

    # check all currently open charts
    for figure_number in plt.get_fignums():
        figure = plt.figure(figure_number)

        # save the chart only once
        if id(figure) not in saved_figures:
            chart_count += 1

            # create a file name for the chart
            file_path = chart_folder / f"chart_{chart_count:02d}.png"

            # save the chart in high quality
            figure.savefig(
                file_path,
                dpi=300,
                bbox_inches="tight"
            )

            saved_figures.add(id(figure))
            saved_figure_objects.append(figure)

    # display the chart normally
    return original_show(*args, **kwargs)

# replace the original show function
plt.show = save_chart_before_display

print("chart export setup is ready!")
```

```python
# Select the numerical columns we want to inspect
numeric_columns = [
    "Weekly_Sales",
    "Temperature",
    "Fuel_Price",
    "CPI",
    "Unemployment"
]

# display basic statistics for each numerical column
print("Descriptive Statistics: ")
display(df[numeric_columns].describe())

# check for unusually low or high values using the IQR method
print("\nOutlier Summary: ")

for column in numeric_columns:
    # calculate the first and third quartiles
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    # calculate the interquartile range
    IQR = Q3 - Q1

    # set the lower and upper limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # count values outside the limits
    outliers = ((df[column] < lower_limit) |
                (df[column] > upper_limit)).sum()

    print(f"{column}: {outliers} potential outliers")
```

```python
# create an empty list to store the results
store_outlier_counts = []

# check weekly sales outliers for each store
for store, store_data in df.groupby("Store"):

    # calculate the first and third quartiles
    Q1 = store_data["Weekly_Sales"].quantile(0.25)
    Q3 = store_data["Weekly_Sales"].quantile(0.75)

    # calculate the interquartile range
    IQR = Q3 - Q1

    # calculate the lower and upper limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # count values outside the allowed range
    outlier_count = (
        (store_data["Weekly_Sales"] < lower_limit) |
        (store_data["Weekly_Sales"] > upper_limit)
    ).sum()

    # add the store and its outlier count to the list
    store_outlier_counts.append({
        "Store": store,
        "Outlier_Count": outlier_count
    })

# convert the results into a DataFrame with defined column names
store_outlier_summary = pd.DataFrame(
    store_outlier_counts,
    columns=["Store", "Outlier_Count"]
)

# display the 10 stores with the most sales outliers
print("Number of stores checked:", len(store_outlier_summary))
display(
    store_outlier_summary
    .sort_values("Outlier_Count", ascending=False)
    .head(10)
)
```

```python
# calculate the global unemployment outlier limits
Q1 = df["Unemployment"].quantile(0.25)
Q3 = df["Unemployment"].quantile(0.75)
IQR = Q3 - Q1

# set the lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# find unemployment values outside the limits
unemployment_outliers = df[
    (df["Unemployment"] < lower_limit) |
    (df["Unemployment"] > upper_limit)
]

# count unemployment outliers for each store
unemployment_summary = (
    unemployment_outliers
    .groupby("Store")["Unemployment"]
    .agg(
        Outlier_Count="count",
        Minimum="min",
        Maximum="max"
    )
    .sort_values("Outlier_Count", ascending=False)
)

# display the 10 stores with the most unemployment outliers
print("Total unemployment outliers:", len(unemployment_outliers))
display(unemployment_summary.head(10))
```

```python
# create an empty list to store the results
unemployment_store_counts = []

# check unemployment values separately for each store
for store, store_data in df.groupby("Store"):

    # calculate the first and third quartiles
    Q1 = store_data["Unemployment"].quantile(0.25)
    Q3 = store_data["Unemployment"].quantile(0.75)

    # calculate the interquartile range
    IQR = Q3 - Q1

    # calculate the lower and upper limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # count values outside the limits
    outlier_count = (
        (store_data["Unemployment"] < lower_limit) |
        (store_data["Unemployment"] > upper_limit)
    ).sum()

    # save the store and its outlier count
    unemployment_store_counts.append({
        "Store": store,
        "Outlier_Count": outlier_count
    })

# convert the results into a DataFrame
unemployment_store_summary = pd.DataFrame(
    unemployment_store_counts,
    columns=["Store", "Outlier_Count"]
)

# calculate the total number of within-store outliers
print(
    "Total within-store unemployment outliers:",
    unemployment_store_summary["Outlier_Count"].sum()
)

# display the stores with the most outliers
display(
    unemployment_store_summary
    .sort_values("Outlier_Count", ascending=False)
    .head(10)
)
```

```python
# group the unemployment values by store
store_groups = df.groupby("Store")["Unemployment"]

# calculate the lower and upper IQR limits for each store
lower_limits = store_groups.transform(
    lambda x: x.quantile(0.25) - 1.5 * (x.quantile(0.75) - x.quantile(0.25))
)

upper_limits = store_groups.transform(
    lambda x: x.quantile(0.75) + 1.5 * (x.quantile(0.75) - x.quantile(0.25))
)

# identify values outside each store's normal range
within_store_outliers = df[
    (df["Unemployment"] < lower_limits) |
    (df["Unemployment"] > upper_limits)
]

# display the store, date, and unemployment value
print("Remaining unemployment outlier records:")
display(
    within_store_outliers[
        ["Store", "Date", "Unemployment"]
    ].sort_values(["Store", "Date"])
)
```

```python
# calculate total and average sales for each store
store_performance = (
    df_prepared.groupby("Store", as_index=False)
    .agg(
        Total_Store_Sales=("Weekly_Sales", "sum"),
        Average_Weekly_Sales=("Weekly_Sales", "mean")
    )
)

# get the 10 stores with the highest total sales
top_10_stores_plot = (
    store_performance
    .nlargest(10, "Total_Store_Sales")
    .copy()
)

# convert Store numbers into categories
top_10_stores_plot["Store"] = (
    top_10_stores_plot["Store"].astype(str)
)

# convert total sales into millions for easier reading
top_10_stores_plot["Sales_Millions"] = (
    top_10_stores_plot["Total_Store_Sales"] / 1_000_000
)

# arrange stores from highest to lowest sales
top_10_stores_plot = top_10_stores_plot.sort_values(
    "Sales_Millions", ascending=False
)

# display the top 10 stores
display(top_10_stores_plot)

# create a horizontal bar chart
plt.figure(figsize=(11, 6))

sns.barplot(
    data=top_10_stores_plot,
    x="Sales_Millions",
    y="Store",
    order=top_10_stores_plot["Store"]
)

# place the highest-performing store at the top
plt.gca().invert_yaxis()

# add chart title and labels
plt.title("Top 10 Walmart Stores by Total Sales")
plt.xlabel("Total Sales (Millions)")
plt.ylabel("Store")

# display the chart
plt.tight_layout()
plt.show()
```

```python
# calculate average sales for holiday and non-holiday weeks
holiday_sales = (
    df_prepared.groupby("Holiday_Flag")["Weekly_Sales"]
    .agg(["mean", "median", "count"])
    .reset_index()
)

# give the holiday flags readable names
holiday_sales["Holiday_Type"] = holiday_sales["Holiday_Flag"].map({
    0: "Non-Holiday",
    1: "Holiday"
})

# display the comparison table
display(holiday_sales)

# create a bar chart to compare average weekly sales
plt.figure(figsize=(8, 5))

sns.barplot(
    data=holiday_sales,
    x="Holiday_Type",
    y="mean"
)

# add the chart title and labels
plt.title("Average Weekly Sales: Holiday vs. Non-Holiday")
plt.xlabel("Week Type")
plt.ylabel("Average Weekly Sales")

# display the chart
plt.tight_layout()
plt.show()
```

```python
# select the numerical columns for correlation analysis
correlation_data = df_prepared[
    [
        "Weekly_Sales",
        "Holiday_Flag",
        "Temperature",
        "Fuel_Price",
        "CPI",
        "Unemployment"
    ]
]

# calculate the correlation between each variable
correlation_matrix = correlation_data.corr()

# create the correlation heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    center=0
)

# add the chart title
plt.title("Correlation Heatmap of Walmart Sales Data")

# display the chart
plt.tight_layout()
plt.show()
```
